L'article *« Robust Medical Image Classification with Uncertain Labels via Evidential Deep Learning »* s'inscrit dans
une ligne de travaux bien établie, dont nous résumons les principes (références dans l'article LaTeX) :

1. **Les labels médicaux sont intrinsèquement incertains.** Désaccords inter-experts, cas ambigus, et labels extraits
   automatiquement des comptes rendus (CheXpert, Irvin et al. 2019, marque explicitement des observations
   « *uncertain* »). Entraîner un classifieur en supposant les labels exacts conduit à mémoriser ce bruit et à des
   prédictions **sur-confiantes**.
2. **Un réseau doit pouvoir dire « je ne sais pas ».** La sortie softmax est une distribution *ponctuelle* : elle ne
   distingue pas « 50/50 parce que les deux classes sont plausibles » de « 50/50 parce que je n'ai jamais vu ça ».
3. **EDL (Sensoy et al., NeurIPS 2018)** : le réseau produit des **évidences** $e_k \ge 0$ qui paramètrent une
   **distribution de Dirichlet** sur les probabilités de classe — une *distribution sur les distributions*. Le manque
   d'évidence devient une masse d'**incertitude** explicite, issue de la théorie de **Dempster-Shafer** via la
   *subjective logic* (Jøsang 2016).
4. **Exploitation clinique** (Ghesu et al., MICCAI 2019 / MedIA 2021, radiographies thoraciques) : l'incertitude
   évidentielle permet de **rejeter** les cas douteux vers un expert (meilleure performance sur les cas conservés) et
   d'**identifier** des exemples d'entraînement probablement mal étiquetés.

**Softmax** : $p_k = e^{z_k}/\sum_j e^{z_j}$, avec $\sum_k p_k = 1$ **toujours** : toute la masse est engagée sur les classes.

**EDL / Dempster-Shafer.** Sur le cadre de discernement $\Omega=\{\text{Normal},\text{Pneumonie}\}$, une *fonction de masse*
peut placer de la masse sur $\Omega$ tout entier — « l'une ou l'autre, je ne sais pas ». En subjective logic, avec $K$ classes :

$$\boxed{\alpha_k = e_k + 1,\qquad S=\sum_{k=1}^K \alpha_k,\qquad b_k = \frac{e_k}{S},\qquad u = \frac{K}{S},\qquad \sum_{k=1}^K b_k + u = 1}$$

* $b_k$ : **croyance** (*belief*) engagée sur la classe $k$ ; $u$ : masse **non engagée** = **ignorance** (masse sur $\Omega$).
* Probabilité attendue : $\hat p_k = \alpha_k / S$ (moyenne de la Dirichlet $\mathrm{Dir}(p\mid\alpha)$).
* Peu d'évidence ($S \to K$) ⇒ $u \to 1$ et $\mathrm{Dir}$ ≈ uniforme : **« je ne sais pas »**, même si $\hat p = (0{,}5;\,0{,}5)$
  comme pour un softmax hésitant. Beaucoup d'évidence équilibrée ⇒ $u$ petit mais $\hat p \approx 0{,}5$ : **ambiguïté réelle**.

**Fonction de perte** (Sensoy et al.) : risque de Bayes de l'entropie croisée sous la Dirichlet + régularisation KL qui
retire l'évidence « trompeuse » (portée par les classes incorrectes) :

$$\mathcal{L}_i = \sum_{k} y_{ik}\big(\psi(S_i) - \psi(\alpha_{ik})\big) \;+\; \lambda_t\,\mathrm{KL}\big(\mathrm{Dir}(p\mid\tilde\alpha_i)\,\|\,\mathrm{Dir}(p\mid\mathbf 1)\big),
\qquad \tilde\alpha_i = y_i + (1-y_i)\odot\alpha_i,\quad \lambda_t=\min(1, t/10)$$

| | Softmax (classique) | EDL (évidentiel) |
|---|---|---|
| Sortie | point $p \in \Delta^{K-1}$ | Dirichlet $\mathrm{Dir}(\alpha)$ sur $\Delta^{K-1}$ |
| Ignorance | non représentable | $u = K/S$ explicite |
| Exemple sans évidence | peut rester très confiant | $u \to 1$ |
| Coût | 1 passe | 1 passe (vs. ensembles / MC-dropout : $T$ passes) |
| Label contradictoire | la CE pousse à mémoriser | le KL pénalise l'évidence non soutenue |

```python
import math, time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import torch
import torch.nn as nn
import torch.nn.functional as F
from torch.utils.data import DataLoader, TensorDataset
from scipy.stats import beta as beta_dist
from sklearn.metrics import roc_auc_score, balanced_accuracy_score

import tp4_utils as U

U.set_seed(); U.setup_plots()
torch.set_num_threads(max(1, torch.get_num_threads()))
data = U.load_pneumonia()
y_train, y_val, y_test = data["y_train"], data["y_val"], data["y_test"]
MEAN, STD = data["img_train"].mean() / 255, data["img_train"].std() / 255

def to_tensor(imgs):
    x = torch.as_tensor(imgs, dtype=torch.float32) / 255.0
    return ((x - MEAN) / STD).unsqueeze(1)          # (n, 1, 28, 28)

Xtr_t, Xva_t, Xte_t = to_tensor(data["img_train"]), to_tensor(data["img_val"]), to_tensor(data["img_test"])
# Jeux hors distribution (OOD) : échographies mammaires et scanners abdominaux (MedMNIST)
breast = U.load_medmnist("breastmnist")["test"][0]
organ = U.load_medmnist("organcmnist")["test"][0]
organ = organ[np.random.default_rng(U.SEED).choice(len(organ), len(y_test), replace=False)]
Xood = {"BreastMNIST (ultrasound)": to_tensor(breast), "OrganCMNIST (CT)": to_tensor(organ)}
print(f"PyTorch {torch.__version__} — {torch.get_num_threads()} threads CPU")
print(f"Train {tuple(Xtr_t.shape)}, test {tuple(Xte_t.shape)}, OOD : " + ", ".join(f"{k}: {len(v)}" for k, v in Xood.items()))
```

```python
class SmallCNN(nn.Module):
    """CNN compact (≈ 25k paramètres), identique pour les deux têtes : seule la sortie/perte change."""
    def __init__(self, n_classes=2):
        super().__init__()
        def block(cin, cout):
            return nn.Sequential(nn.Conv2d(cin, cout, 3, padding=1), nn.BatchNorm2d(cout), nn.ReLU())
        self.features = nn.Sequential(block(1, 16), nn.MaxPool2d(2), block(16, 32), nn.MaxPool2d(2),
                                      block(32, 64), nn.AdaptiveAvgPool2d(1), nn.Flatten())
        self.head = nn.Sequential(nn.Dropout(0.3), nn.Linear(64, n_classes))

    def forward(self, x):
        return self.head(self.features(x))


def kl_to_uniform_dirichlet(alpha):
    """KL( Dir(alpha) || Dir(1,...,1) )."""
    K = alpha.shape[1]
    S = alpha.sum(1, keepdim=True)
    return (torch.lgamma(S).squeeze(1) - math.lgamma(K) - torch.lgamma(alpha).sum(1)
            + ((alpha - 1) * (torch.digamma(alpha) - torch.digamma(S))).sum(1))


def edl_loss(logits, y, epoch, n_classes=2, anneal_epochs=10):
    """Perte évidentielle de Sensoy et al. (2018) : risque de Bayes de la CE + KL recuit."""
    alpha = F.softplus(logits) + 1.0                         # évidence e_k = softplus(z_k) >= 0
    S = alpha.sum(1, keepdim=True)
    Y = F.one_hot(y, n_classes).float()
    risk = (Y * (torch.digamma(S) - torch.digamma(alpha))).sum(1)
    alpha_tilde = Y + (1 - Y) * alpha                        # on ne garde que l'évidence « trompeuse »
    lam = min(1.0, epoch / anneal_epochs)
    return (risk + lam * kl_to_uniform_dirichlet(alpha_tilde)).mean()


def train_cnn(y_fit, head="softmax", seed=0, epochs=15, batch_size=128, lr=1e-3):
    torch.manual_seed(seed)
    model = SmallCNN()
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=1e-4)
    loader = DataLoader(TensorDataset(Xtr_t, torch.as_tensor(y_fit)), batch_size=batch_size, shuffle=True,
                        generator=torch.Generator().manual_seed(seed))
    for epoch in range(epochs):
        model.train()
        for xb, yb in loader:
            logits = model(xb)
            loss = F.cross_entropy(logits, yb) if head == "softmax" else edl_loss(logits, yb, epoch)
            opt.zero_grad(); loss.backward(); opt.step()
    return model.eval()


@torch.no_grad()
def predict(model, X, head):
    """Retourne p(pneumonie), classe prédite et trois mesures d'incertitude."""
    logits = torch.cat([model(X[i:i + 2048]) for i in range(0, len(X), 2048)])
    if head == "softmax":
        p = F.softmax(logits, 1)
        u = None
    else:
        alpha = F.softplus(logits) + 1
        p = alpha / alpha.sum(1, keepdim=True)
        u = (2 / alpha.sum(1)).numpy()                       # u = K / S
    p = p.numpy()
    ent = -(p * np.log(p + 1e-12)).sum(1) / np.log(2)
    out = {"p": p[:, 1], "pred": p.argmax(1), "one_minus_maxp": 1 - p.max(1), "entropy": ent}
    if u is not None:
        out["u"] = u
        out["alpha"] = (F.softplus(logits) + 1).numpy()
    return out

def recalibrated_threshold(p_val):
    """Seuil sur P(pneumonie) maximisant l'accuracy équilibrée sur la validation propre (comme en Partie 3)."""
    cands = np.unique(np.quantile(p_val, np.linspace(0.01, 0.99, 197)))
    return cands[int(np.argmax([balanced_accuracy_score(y_val, (p_val >= t).astype(int)) for t in cands]))]

n_params = sum(p.numel() for p in SmallCNN().parameters())
t0 = time.perf_counter(); _ = train_cnn(y_train, "edl", epochs=1); t_epoch = time.perf_counter() - t0
print(f"SmallCNN : {n_params:,} paramètres — 1 époque ≈ {t_epoch:.1f}s sur CPU")
```

**exactement le même CNN** avec la tête softmax (entropie croisée) ou évidentielle (perte EDL), 15 époques, Adam, 3 graines,
et on évalue sur le test **propre**. Pour chaque modèle on mesure aussi :

* **détection d'erreurs** : AUROC de l'incertitude pour séparer prédictions correctes / incorrectes ;
* **détection hors distribution (OOD)** : AUROC de l'incertitude entre le test PneumoniaMNIST et des images d'une autre
  modalité (échographie mammaire, scanner abdominal) ;
* **détection des labels inversés** dans le train : l'incertitude est-elle plus élevée sur les exemples mal étiquetés ?
* comme en Partie 3, les métriques après **recalibration du seuil** sur la validation propre (`*_recal`).

Incertitude du CNN softmax : entropie normalisée (meilleure des deux mesures classiques, $1-\max p$ donne des AUROC identiques en binaire) ;
incertitude EDL : $u=K/S$.

```python
RATES, SEEDS, EPOCHS = [0.0, 0.05, 0.1, 0.2, 0.3, 0.4], [0, 1, 2], 15
RETRAIN = False      # True : ré-entraîner les 36 CNN (≈ 45-60 min sur CPU) au lieu de relire results/noise_sweep_cnn.csv
MODEL_DIR = U.RESULTS_DIR / "cnn_models"
MODEL_DIR.mkdir(parents=True, exist_ok=True)
model_path = lambda head, rate: MODEL_DIR / f"{head}_noise{int(round(100 * rate))}_seed0.pt"
rows, keep = [], {}
t0 = time.perf_counter()
for rate in (RATES if RETRAIN or not (U.RESULTS_DIR / "noise_sweep_cnn.csv").exists() else []):
    for seed in SEEDS:
        y_fit, flipped = (y_train, np.array([], int)) if rate == 0 else U.flip_minority_labels(y_train, rate, seed=seed)
        for head in ["softmax", "edl"]:
            model = train_cnn(y_fit, head, seed=seed, epochs=EPOCHS)
            te, tr, va = predict(model, Xte_t, head), predict(model, Xtr_t, head), predict(model, Xva_t, head)
            unc = "u" if head == "edl" else "entropy"
            recal = U.binary_metrics(y_test, (te["p"] >= recalibrated_threshold(va["p"])).astype(int))
            r = {"head": head, "noise_rate": rate, "seed": seed, **U.binary_metrics(y_test, te["pred"], te["p"]),
                 **{f"{k}_recal": v for k, v in recal.items()},
                 "ece": U.expected_calibration_error(y_test, te["p"]),
                 "auroc_misclassification": roc_auc_score(te["pred"] != y_test, te[unc])}
            for name, Xo in Xood.items():
                o = predict(model, Xo, head)
                r[f"auroc_ood_{name.split()[0]}"] = roc_auc_score(np.r_[np.zeros(len(y_test)), np.ones(len(Xo))], np.r_[te[unc], o[unc]])
            if len(flipped):
                is_flipped = np.isin(np.arange(len(y_train)), flipped)
                r["auroc_flipped_uncertainty"] = roc_auc_score(is_flipped, tr[unc])
                r["auroc_flipped_disagreement"] = roc_auc_score(is_flipped, np.where(y_fit == 1, 1 - tr["p"], tr["p"]))
            rows.append(r)
            if seed == 0 and rate in (0.0, 0.2):
                keep[(head, rate)] = (model, te)
                torch.save(model.state_dict(), model_path(head, rate))
    print(f"bruit {rate:.0%} terminé — {time.perf_counter() - t0:.0f}s écoulées")
if rows:
    cnn = pd.DataFrame(rows)
    U.save_results("noise_sweep_cnn.csv", cnn)
else:
    cnn = U.load_results("noise_sweep_cnn.csv")
    print("Résultats du balayage relus depuis results/noise_sweep_cnn.csv (RETRAIN = False)")
    for head in ["softmax", "edl"]:
        for rate in (0.0, 0.2):                 # modèles (graine 0) utilisés par les figures d'incertitude
            model = SmallCNN()
            if model_path(head, rate).exists():
                model.load_state_dict(torch.load(model_path(head, rate)))
                model.eval()
            else:
                y_fit = y_train if rate == 0 else U.flip_minority_labels(y_train, rate, seed=0)[0]
                model = train_cnn(y_fit, head, seed=0, epochs=EPOCHS)
                torch.save(model.state_dict(), model_path(head, rate))
            keep[(head, rate)] = (model, predict(model, Xte_t, head))
cnn.groupby(["head", "noise_rate"])[["accuracy", "specificity", "f1", "roc_auc", "ece", "auroc_misclassification"]].mean().round(3)
```

Les chiffres sont dans les tableaux ci-dessus (et `article/tables/tab_edl.tex`, `tab_uncertainty.tex`). Points à retenir :

* **Classification : pas de miracle.** À architecture égale, le CNN évidentiel classe aussi bien que le CNN softmax (AUC
  ≈ 0,94-0,95 dans les deux cas). Les deux CNN souffrent encore plus que les modèles classiques du décalage de prévalence
  train → test (spécificité faible et très variable d'une graine à l'autre au seuil 0,5) ; après **recalibration du seuil**
  sur la validation, tous les modèles se rejoignent autour de 0,86-0,87 d'accuracy.
* **Calibration : l'EDL est plus honnête.** Son erreur de calibration (ECE) est plus faible que celle du softmax sans bruit
  et surtout sous bruit (≈ 0,13 contre ≈ 0,21 à 20 % de labels inversés) : ses probabilités traduisent mieux sa fiabilité réelle.
* **Labels inversés : détectables.** Sur le train bruité à 20 %, l'incertitude classe les exemples mal étiquetés au-dessus
  des autres avec une AUROC ≈ 0,88 pour les deux modèles ; le désaccord entre la prédiction évidentielle et le label donné
  fait encore mieux (≈ 0,92). Nuance importante (histogramme de droite) : les images inversées ont un $u$ élevé **parce
  que ce sont des radiographies normales** — toute la classe « contestée » devient incertaine — et non parce que chaque
  exemple mal étiqueté est isolé parmi ses pairs. C'est un signal que **la régularisation de la Partie 3 ne fournit pas**,
  utile pour cibler la classe à faire ré-annoter plutôt que pour désigner chaque erreur.
* **Hors distribution : résultat mitigé.** Sur ce petit CNN, $u$ ne détecte pas mieux que l'entropie softmax les images
  d'une autre modalité (meilleur sur le scanner à 20 % de bruit, moins bon sur l'échographie). L'EDL n'est pas une garantie
  de détection OOD : cela dépend de l'architecture, de la régularisation KL et des données (cf. Ulmer et al., 2023).
