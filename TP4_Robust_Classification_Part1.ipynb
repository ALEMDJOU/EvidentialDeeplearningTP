```python
n_nan = sum(int(np.isnan(data[k]).sum()) for k in ["X_train", "X_val", "X_test"])
print(f"Valeurs manquantes dans PneumoniaMNIST : {n_nan}")

rng = np.random.default_rng(U.SEED)
def with_missing(X, rate=0.05):
    Xm = X.copy()
    Xm[rng.random(X.shape) < rate] = np.nan
    return Xm

X_train_nan, X_val_nan = with_missing(X_train), with_missing(X_val)
rows = []
for name, Xtr, Xva, imputer in [
    ("Données complètes (référence)", X_train, X_val, None),
    ("Imputation par la moyenne", X_train_nan, X_val_nan, SimpleImputer(strategy="mean")),
    ("Imputation par la médiane", X_train_nan, X_val_nan, SimpleImputer(strategy="median")),
    ("Imputation constante (0)", X_train_nan, X_val_nan, SimpleImputer(strategy="constant", fill_value=0.0)),
]:
    steps = ([imputer] if imputer is not None else []) + [StandardScaler(), LogisticRegression(C=0.01, max_iter=2000)]
    pipe = make_pipeline(*steps).fit(Xtr, y_train)
    rows.append({"Stratégie": name, "Accuracy val": accuracy_score(y_val, pipe.predict(Xva))})
pd.DataFrame(rows).round(4)
```

**Modèle.** La régression logistique modélise la probabilité de la classe positive par la sigmoïde d'un score linéaire :

$$\boxed{h_\theta(x) = P(y=1 \mid x;\theta, b) = \sigma(\theta^\top x + b) = \frac{1}{1 + e^{-(\theta^\top x + b)}}}$$

**Apprentissage par maximum de vraisemblance.** On *maximise* la log-vraisemblance (régularisée L2) :

$$\ell(\theta) = \sum_{i=1}^{n} \Big[ y_i \log h_\theta(x_i) + (1-y_i)\log\big(1-h_\theta(x_i)\big) \Big] - \frac{\lambda}{2}\lVert\theta\rVert^2,
\qquad \frac{\partial \ell_i}{\partial \theta} = \big(y_i - h_\theta(x_i)\big)\,x_i - \lambda\theta$$

**Stochastic Gradient Ascent (SGA).** Au lieu du gradient complet (coût $O(n\,m)$ par mise à jour), on monte le
gradient d'**un seul** exemple tiré au hasard (coût $O(m)$) :

$$\boxed{\theta \leftarrow \theta + \alpha_t \Big[\big(y_i - h_\theta(x_i)\big)\,x_i - \lambda\theta\Big], \qquad \alpha_t = \frac{\alpha_0}{1 + \kappa\, t}}$$

La décroissance $\alpha_t$ (Robbins–Monro) assure la convergence malgré le bruit du gradient stochastique.

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
def sigmoid(z):
    """Sigmoïde numériquement stable."""
    return 0.5 * (1.0 + np.tanh(0.5 * z))


class LogisticRegressionSGA:
    """Régression logistique binaire entraînée par Stochastic Gradient Ascent (implémentation from scratch)."""

    def __init__(self, lr=0.01, decay=1e-4, l2=1e-4, epochs=20, seed=U.SEED):
        self.lr, self.decay, self.l2, self.epochs, self.seed = lr, decay, l2, epochs, seed

    def log_likelihood(self, X, y):
        """Log-vraisemblance moyenne par exemple (à maximiser)."""
        h = np.clip(self.predict_proba(X), 1e-12, 1 - 1e-12)
        return float(np.mean(y * np.log(h) + (1 - y) * np.log(1 - h)))

    def fit(self, X, y, X_val=None, y_val=None):
        rng = np.random.default_rng(self.seed)
        n, m = X.shape
        self.theta, self.bias, t = np.zeros(m), 0.0, 0
        self.history = []
        for epoch in range(self.epochs):
            for i in rng.permutation(n):            # un exemple à la fois (stochastique)
                alpha = self.lr / (1 + self.decay * t)
                error = y[i] - sigmoid(X[i] @ self.theta + self.bias)
                self.theta += alpha * (error * X[i] - self.l2 * self.theta)
                self.bias += alpha * error
                t += 1
            rec = {"epoch": epoch + 1, "train_ll": self.log_likelihood(X, y)}
            if X_val is not None:
                rec["val_ll"] = self.log_likelihood(X_val, y_val)
                rec["val_acc"] = accuracy_score(y_val, self.predict(X_val))
            self.history.append(rec)
        return self

    def decision_function(self, X):
        return X @ self.theta + self.bias

    def predict_proba(self, X):
        return sigmoid(self.decision_function(X))

    def predict(self, X, threshold=0.5):
        return (self.predict_proba(X) >= threshold).astype(int)

print("✅ Classe LogisticRegressionSGA définie")
```

Sélection du pas initial $\alpha_0$ et de la régularisation $\lambda$ sur la **validation** (log-vraisemblance) :

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
grid = []
t0 = time.perf_counter()
for lr in [0.001, 0.003, 0.01]:
    for l2 in [1e-4, 1e-3, 1e-2]:
        mdl = LogisticRegressionSGA(lr=lr, l2=l2, epochs=10).fit(Xtr, y_train)
        grid.append({"alpha0": lr, "lambda": l2, "val_ll": mdl.log_likelihood(Xva, y_val),
                     "val_acc": accuracy_score(y_val, mdl.predict(Xva))})
grid = pd.DataFrame(grid).sort_values("val_ll", ascending=False)
best_sga = grid.iloc[0]
print(f"Recherche sur grille terminée en {time.perf_counter() - t0:.1f}s")
grid.head(5).round(4)
```

La question clé de la Task 1.2 : *pourquoi* le SGA passe à l'échelle. Sur un problème synthétique de
**200 000 exemples**, on compare à budget de temps égal :

* **Batch Gradient Ascent** : gradient exact sur tout le jeu, $O(n\,m)$ par pas (vectorisé NumPy) ;
* **SGA (1 exemple)** : notre implémentation ci-dessus, $O(m)$ par pas ;
* **Mini-batch SGA (256)** : compromis vectorisé, standard en deep learning.

On mesure la log-vraisemblance sur un jeu de validation en fonction du **temps de calcul** (le temps d'évaluation est exclu).

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
Xl, yl = make_classification(n_samples=210_000, n_features=50, n_informative=20, class_sep=0.8, flip_y=0.05, random_state=U.SEED)
Xl = StandardScaler().fit_transform(Xl)
Xl_tr, yl_tr, Xl_va, yl_va = Xl[:200_000], yl[:200_000], Xl[200_000:], yl[200_000:]

def ll(theta, b, X, y):
    h = np.clip(sigmoid(X @ theta + b), 1e-12, 1 - 1e-12)
    return float(np.mean(y * np.log(h) + (1 - y) * np.log(1 - h)))

def run(method, budget=6.0, lr=0.05, batch=256, eval_every=0.15):
    rng = np.random.default_rng(U.SEED)
    n, m = Xl_tr.shape
    theta, b, t, elapsed, next_eval, seen = np.zeros(m), 0.0, 0, 0.0, 0.0, 0
    curve = [(0.0, 0, ll(theta, b, Xl_va, yl_va))]
    while elapsed < budget:
        start = time.perf_counter()
        if method == "batch":
            err = yl_tr - sigmoid(Xl_tr @ theta + b)
            theta += 0.5 * (Xl_tr.T @ err) / n; b += 0.5 * err.mean(); seen += n
        elif method == "sga":
            for i in rng.integers(0, n, 2000):          # 2000 pas stochastiques entre deux évaluations
                a = lr / (1 + 1e-4 * t)
                e = yl_tr[i] - sigmoid(Xl_tr[i] @ theta + b)
                theta += a * e * Xl_tr[i]; b += a * e; t += 1
            seen += 2000
        else:
            for _ in range(20):
                i = rng.integers(0, n, batch)
                e = yl_tr[i] - sigmoid(Xl_tr[i] @ theta + b)
                theta += 0.5 * (Xl_tr[i].T @ e) / batch; b += 0.5 * e.mean()
            seen += 20 * batch
        elapsed += time.perf_counter() - start
        if elapsed >= next_eval:
            curve.append((elapsed, seen / n, ll(theta, b, Xl_va, yl_va))); next_eval += eval_every
    return np.array(curve)

curves = {name: run(key) for name, key in [("Batch gradient ascent", "batch"), ("SGA (1 sample)", "sga"), ("Mini-batch SGA (256)", "minibatch")]}
fig, ax = plt.subplots(1, 2, figsize=(11, 3.6), sharey=True)
for (name, c), col in zip(curves.items(), [U.COLORS["logreg"], U.COLORS["sga"], U.COLORS["svm_lin"]]):
    ax[0].semilogx(np.maximum(c[:, 1], 1e-3), c[:, 2], label=name, lw=2, color=col)
    ax[1].plot(c[:, 0], c[:, 2], label=name, lw=2, color=col)
ax[0].set(xlabel="passes over the 200k training samples (log)", ylabel="validation log-likelihood",
          title="Statistical efficiency", ylim=(-0.60, -0.49))
ax[1].set(xlabel="wall-clock time (s)", title="Computational efficiency (NumPy)")
ax[1].legend()
plt.tight_layout()
U.save_fig(fig, "fig_sga_scaling")
plt.show()
best = max(c[:, 2].max() for c in curves.values())
for name, c in curves.items():
    ok = c[c[:, 2] >= best - 0.005]
    print(f"{name:24s}: LL finale = {c[-1, 2]:.4f} | pour arriver à 0,005 du meilleur LL : "
          + (f"{ok[0, 1]:.3f} passe(s) sur les données, {ok[0, 0]:.2f}s" if len(ok) else "non atteint"))
```

<div style="background:rgba(142,36,170,.07);border-left:4px solid #8e24aa;border-radius:10px;padding:12px 16px;margin:6px 0;font-family:'Segoe UI',system-ui,sans-serif;"><b style="color:#8e24aa;">⚡ Lecture</b><div style="margin-top:6px;line-height:1.55;">
Chaque pas du batch coûte un passage complet sur les 200 000 exemples ; le SGA fait des milliers de mises à jour
(bruitées mais bon marché) dans le même temps et atteint une bonne solution <b>avant même d'avoir vu une fois</b>
toutes les données (panneau de gauche : une fraction de passe contre plusieurs passes complètes pour le batch).
En temps réel (panneau de droite), notre boucle Python « un exemple à la fois » est pénalisée par l'interpréteur, alors
que le batch profite du calcul matriciel optimisé : le <b>mini-batch SGA</b> combine les deux avantages et converge
le plus vite. C'est la raison pour laquelle toute l'optimisation des réseaux profonds (y compris le modèle EDL de la
Partie 4) repose sur des variantes stochastiques du gradient : le coût d'une mise à jour ne dépend pas de $n$.</div></div>

* $\sigma$ envoie le score réel $z=\theta^\top x+b$ dans $]0,1[$ : $h_\theta(x)$ s'interprète comme $P(y=1\mid x)$ (modèle de Bernoulli).
* En inversant : $z = \log\dfrac{P(y=1\mid x)}{P(y=0\mid x)}$ est le **log-odds**. Chaque poids $\theta_j$ est l'augmentation
  du log-odds de pneumonie quand le pixel $j$ (standardisé) augmente d'un écart-type ⇒ la carte des poids est interprétable.
* La probabilité n'est utile que si elle est **calibrée** : parmi les images prédites à 80 %, ~80 % doivent être des pneumonies.
  On le vérifie avec un diagramme de fiabilité et l'*Expected Calibration Error* (ECE).
* Limite (qui motive la Partie 4) : $h_\theta(x)$ ne distingue pas « *je ne sais pas* » de « *c'est 50/50* » : la sigmoïde
  ne représente pas l'**ignorance**.

</div>

La généralisation à $K$ classes remplace la sigmoïde par la **softmax** (régression logistique multinomiale) :

$$\boxed{P(y=k\mid x) = \frac{e^{\theta_k^\top x + b_k}}{\sum_{j=1}^{K} e^{\theta_j^\top x + b_j}}}, \qquad
\theta_k \leftarrow \theta_k + \alpha\,\big(\mathbb{1}[y_i=k] - P(y=k\mid x_i)\big)\,x_i$$

La règle de mise à jour a exactement la même forme « (cible − prédiction) × entrée » que dans le cas binaire.
Nous l'appliquons à **OrganCMNIST** (MedMNIST : coupes CT abdominales 28×28, **11 organes**), en mini-batch SGA.

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
organ = U.load_medmnist("organcmnist")
Xo_tr, yo_tr = U.flatten(organ["train"][0]), organ["train"][1]
Xo_te, yo_te = U.flatten(organ["test"][0]), organ["test"][1]
sc_o = StandardScaler().fit(Xo_tr)
Xo_tr, Xo_te = sc_o.transform(Xo_tr), sc_o.transform(Xo_te)
K = int(yo_tr.max() + 1)
print(f"OrganCMNIST : {len(yo_tr)} train / {len(yo_te)} test, {K} classes")

def softmax(Z):
    Z = Z - Z.max(1, keepdims=True)
    E = np.exp(Z)
    return E / E.sum(1, keepdims=True)

def softmax_sga(X, y, K, lr=0.05, l2=1e-4, epochs=15, batch=64, seed=U.SEED):
    rng = np.random.default_rng(seed)
    W, b, Y = np.zeros((X.shape[1], K)), np.zeros(K), np.eye(K)[y]
    for epoch in range(epochs):
        a = lr / (1 + epoch)
        perm = rng.permutation(len(X))
        for start in range(0, len(X), batch):
            i = perm[start:start + batch]
            G = Y[i] - softmax(X[i] @ W + b)          # (cible - prédiction)
            W += a * (X[i].T @ G / len(i) - l2 * W)
            b += a * G.mean(0)
    return W, b

t0 = time.perf_counter()
W_o, b_o = softmax_sga(Xo_tr, yo_tr, K)
pred_sga_o = (Xo_te @ W_o + b_o).argmax(1)
t_sga = time.perf_counter() - t0
t0 = time.perf_counter()
sk_o = LogisticRegression(C=0.01, max_iter=2000).fit(Xo_tr, yo_tr)
t_sk = time.perf_counter() - t0
print(f"Softmax-SGA (from scratch) : accuracy test = {accuracy_score(yo_te, pred_sga_o):.4f}  ({t_sga:.1f}s)")
print(f"sklearn multinomial        : accuracy test = {accuracy_score(yo_te, sk_o.predict(Xo_te)):.4f}  ({t_sk:.1f}s)")

fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay(confusion_matrix(yo_te, pred_sga_o, normalize="true")).plot(ax=ax, cmap="Purples", values_format=".2f", colorbar=False)
ax.set_title("OrganCMNIST — softmax regression (SGA), row-normalised")
for t in ax.texts: t.set_fontsize(6)
plt.tight_layout(); plt.show()
```

<div style="background:linear-gradient(135deg,rgba(142,36,170,.09),rgba(123,31,162,.04));border-left:5px solid #8e24aa;border-radius:12px;padding:4px 20px 10px 20px;margin:8px 0;box-shadow:0 4px 16px rgba(142,36,170,.10);font-family:'Segoe UI',system-ui,sans-serif;line-height:1.6;">
<div style="height:3px;margin:0 -20px 6px -20px;border-radius:3px;background:linear-gradient(90deg,#f0f0f0,#8e24aa,#f0f0f0);background-size:300% 100%;animation:tp4Flow 7s ease-in-out infinite;"></div>

$$\boxed{P(y\mid x) = \frac{P(y)\prod_{j=1}^{m} P(x_j\mid y)}{P(x)}}
\qquad\text{(Gaussien : } P(x_j\mid y=k) = \mathcal{N}(x_j;\ \mu_{jk}, \sigma^2_{jk})\text{)}$$

**Hypothèse d'indépendance conditionnelle** : sachant la classe, les features sont indépendantes.

* **Justification / avantages** : au lieu d'estimer une loi jointe sur $m=784$ dimensions (une covariance pleine a
  $m(m+1)/2 \approx 3{,}1\times10^5$ paramètres par classe), on estime $2m$ paramètres par classe. L'entraînement est
  un simple calcul de moyennes et variances en **$O(n\,m)$, une seule passe**, et la prédiction est en $O(m\,K)$.
  Avec peu de données, ce biais fort réduit énormément la variance de l'estimateur. Le classifieur reste souvent bon
  car seule la **position de la frontière** compte, pas l'exactitude des probabilités (Domingos & Pazzani, 1997).
* **Limite** : sur des images, l'hypothèse est clairement **fausse** — des pixels voisins sont très corrélés. On le mesure
  ci-dessous, puis on teste une projection **PCA** (features décorrélées) pour voir si une hypothèse « plus vraie » aide.

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
corr_center = np.array([np.corrcoef(X_train[:, 14 * 28 + 14], X_train[:, j])[0, 1] for j in range(784)])
adj = np.mean([np.corrcoef(X_train[:, r * 28 + c], X_train[:, r * 28 + c + 1])[0, 1] for r in range(4, 24) for c in range(4, 23)])
print(f"Corrélation moyenne entre pixels horizontalement voisins (zone centrale) : {adj:.3f}")

t0 = time.perf_counter(); nb = GaussianNB().fit(Xtr, y_train); t_nb = time.perf_counter() - t0
pca50 = PCA(n_components=50, random_state=U.SEED).fit(Xtr)
nb_pca = GaussianNB().fit(pca50.transform(Xtr), y_train)
t0 = time.perf_counter(); LogisticRegression(C=best_C_logreg, max_iter=3000).fit(Xtr, y_train); t_lr = time.perf_counter() - t0

print(f"Temps d'entraînement : Naive Bayes = {t_nb*1000:.1f} ms | Régression logistique (L-BFGS) = {t_lr*1000:.0f} ms")
print(f"Accuracy val — GaussianNB (784 pixels) : {accuracy_score(y_val, nb.predict(Xva)):.4f}")
print(f"Accuracy val — GaussianNB (50 composantes PCA, décorrélées) : {accuracy_score(y_val, nb_pca.predict(pca50.transform(Xva))):.4f}")

fig, ax = plt.subplots(figsize=(3.6, 3.2))
im = ax.imshow(corr_center.reshape(28, 28), cmap="magma", vmin=0, vmax=1)
ax.plot(14, 14, "c+", ms=10, mew=2); ax.axis("off")
ax.set_title("corr(central pixel, every pixel)")
plt.colorbar(im, ax=ax, fraction=0.046); plt.show()
print(f"AUC val — NB pixels : {roc_auc_score(y_val, nb.predict_proba(Xva)[:, 1]):.3f} | "
      f"NB + PCA : {roc_auc_score(y_val, nb_pca.predict_proba(pca50.transform(Xva))[:, 1]):.3f}")
```

<div style="background:rgba(142,36,170,.07);border-left:4px solid #8e24aa;border-radius:10px;padding:12px 16px;margin:6px 0;font-family:'Segoe UI',system-ui,sans-serif;"><b style="color:#8e24aa;">🔍 Ce que montre l'expérience</b><div style="margin-top:6px;line-height:1.55;">
Des pixels voisins ont une corrélation de l'ordre de 0,9 : l'hypothèse d'indépendance est massivement violée, et NB
« compte » plusieurs fois la même information, ce qui rend ses probabilités <b>extrêmes et mal calibrées</b>.
Rendre l'hypothèse « plus vraie » par PCA <b>n'aide pas</b> ici sur la validation (accuracy et AUC plus faibles) :
non-corrélation n'est pas indépendance, et les 50 composantes ne sont pas gaussiennes conditionnellement à la classe.
Ce qui compte pour NB n'est pas que l'hypothèse soit vraie, mais que la frontière qu'elle induit soit bonne
(Domingos & Pazzani). NB reste de loin le modèle le plus rapide à entraîner : une excellente <i>baseline</i>, pas un modèle final.</div></div>

<div style="margin-top:6px;"><span style="display:inline-block;background:linear-gradient(90deg,#ffffff,#4facfe,#0006ab);background-size:150% 100%;-webkit-background-clip:text;background-clip:text;color:transparent;-webkit-text-fill-color:transparent;font-family:'Montserrat','Segoe UI',sans-serif;font-size:1.55rem;font-weight:800;animation:tp4Flow 7s ease-in-out infinite;">📍 Task 1.3 : K plus proches voisins (KNN)</span></div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

<div style="background:linear-gradient(135deg,rgba(142,36,170,.09),rgba(123,31,162,.04));border-left:5px solid #8e24aa;border-radius:12px;padding:4px 20px 10px 20px;margin:8px 0;box-shadow:0 4px 16px rgba(142,36,170,.10);font-family:'Segoe UI',system-ui,sans-serif;line-height:1.6;">
<div style="height:3px;margin:0 -20px 6px -20px;border-radius:3px;background:linear-gradient(90deg,#f0f0f0,#8e24aa,#f0f0f0);background-size:300% 100%;animation:tp4Flow 7s ease-in-out infinite;"></div>

$$\hat y(x) = \operatorname{majorité}\{\,y_i : x_i \in \mathcal{N}_K(x)\,\}, \qquad d(a,b) = \sqrt{\textstyle\sum_{j}(a_j-b_j)^2}$$

* **Rôle de $K$** : $K$ petit ⇒ frontière très irrégulière, faible biais / forte variance (sensible au **bruit de labels** :
  un seul voisin mal étiqueté suffit à se tromper) ; $K$ grand ⇒ frontière lisse, plus de biais. $K$ est choisi sur la validation.
* **Malédiction de la dimensionnalité** : en grande dimension, les distances se **concentrent** — le plus proche et le plus
  lointain voisin sont presque à la même distance, et la notion de « voisin » perd son sens ; il faudrait un nombre
  d'exemples exponentiel en $m$ pour couvrir l'espace. KNN n'apprend rien (*lazy learner*) : chaque prédiction coûte
  $O(n\,m)$ et tout le jeu d'entraînement doit être stocké.

</div>
<style>@keyframes tp4Flow{0%,100%{background-position:0% 50%}50%{background-position:100% 50%}}@keyframes tp4Glow{0%,100%{filter:brightness(1)}50%{filter:brightness(1.18)}}</style>

```python
k_max = 61
knn_full = KNeighborsClassifier(n_neighbors=k_max).fit(Xtr, y_train)
_, neigh = knn_full.kneighbors(Xva)                    # une seule recherche, puis vote pour chaque K
k_grid = list(range(1, k_max + 1, 2))
knn_val = pd.DataFrame([{"K": k, "val_acc": accuracy_score(y_val, (y_train[neigh[:, :k]].mean(1) > 0.5).astype(int))} for k in k_grid])
best_k = int(knn_val.sort_values("val_acc", ascending=False).iloc[0].K)
print(f"Meilleur K (validation) : {best_k}")

# (a) concentration des distances sur données uniformes
rng = np.random.default_rng(U.SEED)
dims = [2, 5, 10, 20, 50, 100, 200, 500, 784]
contrast = []
for d in dims:
    P, q = rng.random((2000, d)), rng.random(d)
    dist = np.linalg.norm(P - q, axis=1)
    contrast.append((dist.max() - dist.min()) / dist.min())
# (b) KNN sur PneumoniaMNIST en fonction de la dimension (PCA)
pca_full = PCA(random_state=U.SEED).fit(Xtr)
pca_dims = [2, 5, 10, 20, 50, 100, 200, 400, 784]
acc_dim = []
for d in pca_dims:
    Ztr, Zva = pca_full.transform(Xtr)[:, :d], pca_full.transform(Xva)[:, :d]
    acc_dim.append(accuracy_score(y_val, KNeighborsClassifier(n_neighbors=best_k).fit(Ztr, y_train).predict(Zva)))

fig, ax = plt.subplots(1, 3, figsize=(13, 3.4))
ax[0].plot(knn_val.K, knn_val.val_acc, "o-", ms=3, color=U.COLORS["knn"], lw=2)
ax[0].axvline(best_k, ls="--", color="grey"); ax[0].set(xlabel="K", ylabel="validation accuracy", title="Role of K")
ax[1].loglog(dims, contrast, "s-", color=U.COLORS["knn"], lw=2)
ax[1].set(xlabel="dimension d", ylabel=r"$(d_{max}-d_{min})/d_{min}$", title="Distance concentration (uniform data)")
ax[2].semilogx(pca_dims, acc_dim, "o-", color=U.COLORS["knn"], lw=2)
ax[2].set(xlabel="number of PCA components", ylabel="validation accuracy", title=f"KNN (K={best_k}) vs dimension")
plt.tight_layout()
U.save_fig(fig, "fig_knn")
plt.show()
print("Accuracy val de KNN selon le nombre de composantes PCA :", {d: round(float(a), 4) for d, a in zip(pca_dims, acc_dim)})
print(f"Variance expliquée par 50 composantes : {pca_full.explained_variance_ratio_[:50].sum():.1%}")
```

<div style="background:rgba(142,36,170,.07);border-left:4px solid #8e24aa;border-radius:10px;padding:12px 16px;margin:6px 0;font-family:'Segoe UI',system-ui,sans-serif;"><b style="color:#8e24aa;">🔍 Interprétation</b><div style="margin-top:6px;line-height:1.55;">
Sur des données uniformes, le contraste des distances s'effondre avec la dimension : c'est la malédiction de la
dimensionnalité. Pourtant KNN fonctionne bien sur les 784 pixels : les radiographies ne remplissent pas l'espace, elles
vivent sur une <b>variété de faible dimension intrinsèque</b> (quelques dizaines de composantes captent l'essentiel de la
variance), et c'est cette dimension intrinsèque qui gouverne la difficulté. Au-delà de ~100 composantes, l'accuracy
plafonne : réduire la dimension ne dégrade presque rien et divise le coût de chaque prédiction. Le petit $K$ optimal
rend en revanche KNN <b>fragile face aux labels bruités</b> (Partie 3).</div></div>
